## Dataset Research — AgriDataValue

Dataset yang sedang dianalisis:

**Environmental Data for Powdery Mildew, Downy Mildew**

Dataset berisi data lingkungan berbasis IoT pada tanaman grapevine/vineyard, termasuk variabel:

- Air temperature
- Minimum/maximum air temperature
- Air humidity
- Minimum/maximum air humidity
- Precipitation/pluviometer
- Downy mildew
- Powdery mildew
- Botrytis
- Erysiphe

Dataset menyediakan beberapa bentuk data:

1. **Original data** — data observasi utama
2. **Noised data** — variasi data dengan penambahan noise
3. **Synthetic data** — data sintetis
4. **Synthetic + noised data** — data sintetis yang diberi noise

Untuk **baseline research**, analisis utama menggunakan **original data** terlebih dahulu. Data noised dan synthetic digunakan sebagai eksperimen tambahan apabila diperlukan, misalnya untuk menguji robustness model dan penggunaan synthetic data.

````python
import pandas as pd
from pathlib import Path
````

````python
##### DATASET 1
data_dir = Path("../data/data_1")

# Ambil hanya file ORIGINAL
original_files = sorted(data_dir.glob("*original*.csv"))

print("File yang ditemukan:")
for file in original_files:
    print("-", file.name)

# Baca dan gabungkan
dfs = []

for file in original_files:
    temp = pd.read_csv(file, sep="\t")
    
    # Simpan informasi sumber file
    temp["source_file"] = file.name
    
    dfs.append(temp)

# Gabungkan semua tahun
df = pd.concat(dfs, ignore_index=True)

# Hapus kolom index lama kalau memang ada
if "Unnamed: 0" in df.columns:
    df = df.drop(columns=["Unnamed: 0"])

# Konversi tanggal
df["date"] = pd.to_datetime(df["date"])

# Urutkan berdasarkan tanggal
df = df.sort_values("date").reset_index(drop=True)

# Buat ID observasi baru
df.insert(0, "record_id", range(1, len(df) + 1))

# Tampilkan
df
````
