## Dataset Information

**Dataset:**  
*Dataset for: Smart Multidepth Soil Sensing System with an Embedded Hydrothermal Root Zone Stress Index (HRZSI) in Hardpan-Limited Tropical Spodosol Oil Palm Systems*

**Source:** Zenodo

**DOI:** `10.5281/zenodo.20092399`

**Location:** Central Kalimantan, Indonesia

**Crop/System:** Oil Palm

**Monitoring Period:** 225 consecutive days

**Dataset Type:** Environmental / Soil Sensor / Hydrothermal Root-Zone Data

### Main Environmental Variables

Dataset ini mencakup beberapa kelompok variabel:

- Air temperature
- Relative humidity
- Rainfall
- Solar irradiance
- Soil moisture pada beberapa kedalaman
- Soil temperature pada beberapa kedalaman
- Hydrothermal Root Zone Stress Index (HRZSI)
- HRZSI-derived variables dan regime information

```python
import pandas as pd
from pathlib import Path
```

```python
excel_file = pd.ExcelFile(file_path)

print("Jumlah sheet:", len(excel_file.sheet_names))
print("\nNama sheet:")

for i, sheet in enumerate(excel_file.sheet_names, start=1):
    print(f"{i}. {sheet}")
```
