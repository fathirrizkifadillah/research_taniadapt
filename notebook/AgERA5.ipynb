```python
import cdsapi
from pathlib import Path

dataset = "sis-agrometeorological-indicators"
OUT = Path("../data/data1/raw/agera5")          # sesuaikan dengan struktur folder lo
OUT.mkdir(parents=True, exist_ok=True)

YEARS = [2025]                            # UJI DULU 1 tahun. Nanti ganti: range(2010, 2026)
AREA = [-5.8, 106.3, -7.9, 109]           # [N, W, S, E] Jawa Barat
MONTHS = [f"{m:02d}" for m in range(1, 13)]
DAYS = [f"{d:02d}" for d in range(1, 32)]

# (nama file, variable, statistic atau None)
JOBS = [
    ("precip",   "precipitation_flux", None),
    ("tmax",     "2m_temperature", "24_hour_maximum"),
    ("tmin",     "2m_temperature", "24_hour_minimum"),
    ("tmean",    "2m_temperature", "24_hour_mean"),
    # Aktifkan di bawah ini SETELAH 4 di atas berhasil.
    # Nama variabel/statistik di bawah harus dicek dulu lewat "Show API request" di form CDS
    # ("rh_mean",  "2m_relative_humidity_derived", "24_hour_mean"),
    # ("et0",      "reference_evapotranspiration", None),
    # ("vpd",      "vapour_pressure_deficit_at_maximum_temperature", None),
    # ("precdur",  "precipitation_duration_fraction", None),
    # ("srad",     "solar_radiation_flux", "24_hour_mean"),
    # ("wind",     "10m_wind_speed", "24_hour_mean"),
]

client = cdsapi.Client()
gagal = []

for name, variable, statistic in JOBS:
    for year in YEARS:
        target = OUT / f"{name}_{year}.zip"
        if target.exists():
            print(f"skip   {target.name}")
            continue
        request = {
            "variable": variable,
            "year": [str(year)],          
            "month": MONTHS,
            "day": DAYS,
            "version": "2_0",
            "area": AREA,
        }
        if statistic:
            request["statistic"] = [statistic]  
        try:
            print(f"unduh  {target.name} ...")
            client.retrieve(dataset, request).download(str(target))
        except Exception as e:
            print(f"GAGAL  {target.name}: {str(e)[:150]}")
            gagal.append(target.name)

print("\nSelesai. Gagal:", gagal if gagal else "tidak ada")
```

```python
import zipfile, glob
import numpy as np
import pandas as pd
import xarray as xr
import matplotlib.pyplot as plt
from pathlib import Path

RAW = Path("../data/data1/raw/agera5")
FIG = Path("../reports/figures"); FIG.mkdir(parents=True, exist_ok=True)
LAT, LON = -7.33, 108.22

def save_table_png(df, path, title=None):
    fmt = lambda v: f"{v:.1f}" if isinstance(v, (float, np.floating)) else str(v)
    cells = [[fmt(v) for v in row] for row in df.values]
    fig, ax = plt.subplots(figsize=(max(6, 1.4 * len(df.columns)), 0.5 * (len(df) + 2)))
    ax.axis("off")
    tbl = ax.table(cellText=cells, colLabels=list(df.columns), loc="center", cellLoc="center")
    tbl.auto_set_font_size(False); tbl.set_fontsize(10); tbl.scale(1, 1.5)
    for (r, c), cell in tbl.get_celld().items():
        if r == 0:
            cell.set_text_props(weight="bold"); cell.set_facecolor("#e8eef5")
    if title: ax.set_title(title, fontweight="bold", pad=12)
    fig.savefig(path, dpi=200, bbox_inches="tight"); plt.close(fig)
    print("tersimpan:", path)

for z in RAW.glob("*_2025.zip"):
    if not (RAW / z.stem).exists():
        zipfile.ZipFile(z).extractall(RAW / z.stem)

def titik(nama):
    nc = sorted(glob.glob(str(RAW / f"{nama}_2025" / "*.nc")))
    ds = xr.open_mfdataset(nc, combine="by_coords")
    v = list(ds.data_vars)[0]
    s = ds[v].sel(lat=LAT, lon=LON, method="nearest")
    print(f"{nama:7s} var={v}  satuan={ds[v].attrs.get('units')}  hari={s.sizes['time']}  "
          f"min={float(s.min()):.2f} maks={float(s.max()):.2f} rata2={float(s.mean()):.2f}  NaN={int(s.isnull().sum())}")
    return s.to_series()

hujan = titik("precip")
tmax, tmin, tmean = titik("tmax"), titik("tmin"), titik("tmean")

if tmean.mean() > 100:
    tmax, tmin, tmean = tmax - 273.15, tmin - 273.15, tmean - 273.15

# --- grafik harian ---
fig, ax = plt.subplots(2, 1, figsize=(10, 6), sharex=True)
hujan.plot(ax=ax[0], color="tab:blue");  ax[0].set_ylabel("Hujan (mm/hari)")
tmax.plot(ax=ax[1], label="Tmax"); tmean.plot(ax=ax[1], label="Tmean"); tmin.plot(ax=ax[1], label="Tmin")
ax[1].set_ylabel("Suhu (°C)"); ax[1].legend()
plt.suptitle("AgERA5 2025 - titik Tasikmalaya"); plt.tight_layout()
fig.savefig(FIG / "agera5_2025_tasikmalaya_harian.png", dpi=200, bbox_inches="tight")
plt.show()

# --- tabel bulanan ---
bulanan = pd.DataFrame({
    "hujan_mm": hujan.resample("ME").sum().round(0),
    "tmax_C":   tmax.resample("ME").mean().round(1),
    "tmean_C":  tmean.resample("ME").mean().round(1),
    "tmin_C":   tmin.resample("ME").mean().round(1),
})
bulanan.index = bulanan.index.strftime("%Y-%m")
bulanan = bulanan.reset_index().rename(columns={"time": "bulan"})
print(bulanan)

save_table_png(bulanan, FIG / "agera5_2025_tasikmalaya_bulanan.png",
               title="AgERA5 2025 - Tasikmalaya, ringkasan bulanan")
bulanan.to_csv(FIG / "agera5_2025_tasikmalaya_bulanan.csv", index=False)

print("\nTotal hujan 2025:", round(hujan.sum()), "mm")
print("Tanggal hujan maksimum:", hujan.idxmax().date(), "=", round(hujan.max(), 1), "mm")
```

```python
import glob
import numpy as np
import xarray as xr
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

BASE = "../data/data1/raw/agera5"
FIG = Path("../reports/figures"); FIG.mkdir(parents=True, exist_ok=True)

def save_table_png(df, path, title=None):
    fmt = lambda v: f"{v:.1f}" if isinstance(v, (float, np.floating)) else str(v)
    cells = [[fmt(v) for v in row] for row in df.values]
    fig, ax = plt.subplots(figsize=(max(6, 1.4 * len(df.columns)), 0.5 * (len(df) + 2)))
    ax.axis("off")
    tbl = ax.table(cellText=cells, colLabels=list(df.columns), loc="center", cellLoc="center")
    tbl.auto_set_font_size(False); tbl.set_fontsize(10); tbl.scale(1, 1.5)
    for (r, c), cell in tbl.get_celld().items():
        if r == 0:
            cell.set_text_props(weight="bold"); cell.set_facecolor("#e8eef5")
    if title: ax.set_title(title, fontweight="bold", pad=12)
    fig.savefig(path, dpi=200, bbox_inches="tight"); plt.close(fig)
    print("tersimpan:", path)

def open_var(name, year=2025):
    files = sorted(glob.glob(f"{BASE}/{name}_{year}/**/*.nc", recursive=True))
    assert files, f"Tidak ada .nc di {BASE}/{name}_{year}"
    return xr.open_mfdataset(files, combine="by_coords").load()   # .load() = jauh lebih cepat

ds = {k: open_var(k) for k in ["precip", "tmax", "tmean", "tmin"]}
vname = {k: list(v.data_vars)[0] for k, v in ds.items()}
print("variabel:", vname)

latn = "lat" if "lat" in ds["precip"].coords else "latitude"
lonn = "lon" if "lon" in ds["precip"].coords else "longitude"

points = {
    "Tasikmalaya": (-7.33, 108.22),
    "Bandung":     (-6.91, 107.61),
    "Karawang":    (-6.30, 107.30),
    "Pangandaran": (-7.69, 108.65),
}

rows = []
for name, (la, lo) in points.items():
    sel = {k: ds[k][vname[k]].sel({latn: la, lonn: lo}, method="nearest") for k in ds}
    cell_lat = float(sel["precip"][latn]); cell_lon = float(sel["precip"][lonn])
    g = {k: s.to_series() for k, s in sel.items()}
    df = pd.DataFrame({"pr": g["precip"], "tx": g["tmax"] - 273.15,
                       "tm": g["tmean"] - 273.15, "tn": g["tmin"] - 273.15})
    rows.append({
        "titik": name,
        "sel_lat": round(cell_lat, 2), "sel_lon": round(cell_lon, 2),
        "hujan_mm": int(round(float(df.pr.sum()))),
        "hari>=1mm": int((df.pr >= 1).sum()),
        "hari>=50mm": int((df.pr >= 50).sum()),
        "tmax": round(float(df.tx.mean()), 1),
        "tmean": round(float(df.tm.mean()), 1),
        "tmin": round(float(df.tn.mean()), 1),
        "urutan_ok_%": round(float(((df.tn <= df.tm) & (df.tm <= df.tx)).mean() * 100), 1),
    })

tabel = pd.DataFrame(rows)
print(tabel.to_string(index=False))

save_table_png(tabel, FIG / "agera5_2025_multititik.png",
               title="AgERA5 2025 - perbandingan 4 titik Jawa Barat")
tabel.to_csv(FIG / "agera5_2025_multititik.csv", index=False)
```
